```python
query = """
SELECT
    Payment_Method,
    ROUND(SUM(Quantity * Unit_Price * (1 - Discount)), 2) AS Total_Revenue
FROM sales
GROUP BY Payment_Method
ORDER BY Total_Revenue DESC;
"""

pd.read_sql_query(query, conn)
```

```python
query = """
SELECT
    SUBSTR(Order_Date, 1, 7) AS Month,
    ROUND(SUM(Quantity * Unit_Price * (1 - Discount)), 2) AS Total_Revenue
FROM sales
GROUP BY SUBSTR(Order_Date, 1, 7)
ORDER BY Month;
"""

pd.read_sql_query(query, conn)
```
